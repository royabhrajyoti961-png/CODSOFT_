```python
from sklearn.model_selection import train_test_split

# 1. Separate features (X) and target label (y) using your active 'data_file' DataFrame
X = data_file.drop(columns='species')
y = data_file['species']

# 2. Split into training and testing sets (using your line of code)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
LR=LogisticRegression()
```

```python
from sklearn.linear_model import LogisticRegression

# 1. Instantiate the Logistic Regression model
# max_iter is increased to 200 to ensure the model converges on the Iris dataset
LR = LogisticRegression(max_iter=200)

# 2. Fit the model using your split training data
# Note: If your variables are capitalized as X_train, use capital X here
LR.fit(X_train, y_train)
```

```python
KNN=KNeighborsClassifier()
```

```python
from sklearn.neighbors import KNeighborsClassifier

# 1. Instantiate the KNN classifier (default is usually 5 neighbors)
KNN = KNeighborsClassifier(n_neighbors=5)

# 2. Fit the model using your split training data
# Note: Use capital 'X_train' if that's how you named it in your split cell
KNN.fit(X_train, y_train)
```

```python

DT=DecisionTreeClassifier()
```

```python
from sklearn.tree import DecisionTreeClassifier

# 1. Instantiate the Decision Tree Classifier
DT = DecisionTreeClassifier(random_state=42)

# 2. Fit the model using your split training data
# Note: Use capital 'X_train' if that's how you named it in your split cell
DT.fit(X_train, y_train)
```

```python
# Ensure variable names match your split (use X_test if capitalized)
LR_accuracy = LR.score(X_test, y_test) * 100
KNN_accuracy = KNN.score(X_test, y_test) * 100
DT_accuracy = DT.score(X_test, y_test) * 100

# Print the comparison results cleanly
print(f"Logistic Regression Accuracy: {LR_accuracy:.2f}%")
print(f"KNN Accuracy:                 {KNN_accuracy:.2f}%")
print(f"Decision Tree Accuracy:       {DT_accuracy:.2f}%")
```

```python
import matplotlib.pyplot as plt

# 1. Clear layout setup
fig, ax = plt.subplots(figsize=(6, 1.2))

# 2. Plot background track (100%) and the performance fill
ax.barh(['Decision Tree'], [100], color='#eaedf2', height=0.4, label='Total')
ax.barh(['Decision Tree'], [DT_accuracy], color='#e74c3c', height=0.4, label='Accuracy')

# 3. Annotate the precision metric explicitly
ax.text(DT_accuracy + 1.5, 0, f'{DT_accuracy:.2f}%', va='center', ha='left', fontweight='bold', fontsize=12, color='#2c3e50')

# 4. Clean visual borders
ax.set_xlim(0, 100)
ax.axis('off')

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Consolidate training data metrics
metrics_df = pd.DataFrame({
    'Model': ['Logistic Regression', 'KNN', 'Decision Tree'],
    'Accuracy (%)': [LR_accuracy, KNN_accuracy, DT_accuracy]
})

# 2. Initialize figure structure
plt.figure(figsize=(7, 4))
sns.set_theme(style="whitegrid")

# 3. Plot evaluation metrics
ax = sns.barplot(x='Model', y='Accuracy (%)', data=metrics_df, hue='Model', palette='Set2', legend=False)

# 4. Global labels and bounds adjustment
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%', fontweight='bold', padding=3)

plt.ylim(0, 115)
plt.title('Final Model Accuracy Evaluation Dashboard', fontsize=12, fontweight='bold', pad=15)
plt.tight_layout()
plt.show()
```
